```python
from torch.utils.data import Dataset, DataLoader
import torch
from pathlib import Path
import numpy as np
```

```python

max_per_class = 10
split = "train"
quickdraw_folder = "quickdraw_data"
data_folder = Path(quickdraw_folder)
# Exclude .full.npz files to avoid duplicates
all_files = data_folder.glob("*.npz")
files = sorted([f for f in all_files if not f.stem.endswith('.full')])
split = split
max_per_class = max_per_class
label_to_idx = {f.stem: i for i, f in enumerate(files)}
print(label_to_idx)
# Build index mapping: global_idx -> (file_path, drawing_idx_in_file, label)
index_map = []
n_files_loaded = 0
print(len(files))
for file_path in files:
    if n_files_loaded > 1:
        break
    n_files_loaded +=1
    label = file_path.stem
    # Only load metadata to get count, not actual data
    data = np.load(file_path, allow_pickle=True, encoding='latin1')
    num_drawings = min(len(data[split]), max_per_class)
    
    # Store references, not actual drawings
    for drawing_idx in range(num_drawings):
        index_map.append((file_path, drawing_idx, label_to_idx[label]))
        print(index_map)
    print(n_files_loaded)
    data.close()  # Close the file

print(f"Dataset initialized with {len(index_map)} samples from {len(files)} classes")
```

```python
idx = 3
file_path, drawing_idx, label = index_map[idx]

# Lazy load: only load the specific drawing when needed
data = np.load(file_path, allow_pickle=True, encoding='latin1')
drawing = data[split][drawing_idx]
data.close()

# Convert drawing to tensor
t = torch.tensor(drawing, dtype=torch.float32), label
```
